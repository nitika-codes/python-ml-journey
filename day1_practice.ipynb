```python
coercivity_values = [8.2, 11.5, 9.4, 14.1, 12.0, 7.8, 15.3, 10.1, 6.9, 13.5]
for value in coercivity_values[:5]:
    if value>10.0:
        print(f"{value} kOe: High performance threshold met!")
    else:
        print(f"{value} kOe: Standard grade.")
mid_batch=coercivity_values[3:7]
for values in mid_batch:
    print(values*1.1)
for value in reversed(coercivity_values[-1:-4:-1]):
    print(value)
```
